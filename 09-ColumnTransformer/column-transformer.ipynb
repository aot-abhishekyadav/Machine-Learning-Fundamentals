```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

# 1. Aam Zindagi
without column transformer

```python
# adding simple imputer to fever col
si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[['fever']])
X_test_fever = si.transform(X_test[['fever']])
X_train_fever.shape
```

```python
oe = OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough = oe.fit_transform(X_train[['cough']])
X_test_cough = oe.transform(X_test[['cough']])
X_train_cough.shape
```

```python
ohe = OneHotEncoder(drop='first', dtype=np.int32, sparse_output= False)
X_train_gender_city = ohe.fit_transform(X_train[['gender', 'city']])
X_test_gender_city = ohe.transform(X_test[['gender', 'city']])
X_train_gender_city.shape
```

```python
# Extracting Age
X_train_age = X_train.drop(columns=['gender','fever','cough','city']).values

# also the test data
X_test_age = X_test.drop(columns=['gender','fever','cough','city']).values

X_train_age.shape
```

```python
X_train_new = np.concatenate((X_train_age, X_train_fever, X_train_gender_city, X_train_cough), axis=1)
X_test_new = np.concatenate((X_test_age, X_test_fever, X_test_gender_city, X_test_cough), axis=1)
```

```python
X_train_new.shape
```

# Mentos Zindagi
using column Transformer


```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1', SimpleImputer(), ['fever']),
    ('tnf2', OrdinalEncoder(categories=[['Mild','Strong']]), ['cough']),
    ('tnf3', OneHotEncoder(drop='first', dtype=np.int32, sparse_output= False), ['gender', 'city'])
], remainder='passthrough') # remainder mean agr koi column hai jiss sai hum drop karna chahte hai to drop kar dai gai remainder agr nhi karna chahte hai to passthrough kar dai gai
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```
