# Working With Dates

```python
# converting date into datetime datatype

date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

### 1. Extract Year

```python
date['date_year'] = date['date'].dt.year  # using dt.Year we can extract Year

date.head()
```

### 2. Extract Month

```python
date['date_month_no'] = date['date'].dt.month  # using dt.month we extract month

date.head()
```

3. Extract Month Name

```python
date['date_month_name'] = date['date'].dt.month_name()

date.head()
```

4. Extract Days

```python
date['date_day'] = date['date'].dt.day

date.head()
```

5. name of the Day

```python
date['date_dow_name'] = date['date'].dt.day_name()

date.head()
```

6. Day Number

```python
date['date_week'] = date['date'].dt.dayofweek  # 0 . monday

date.head()
```

7. is weekend

```python
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

8. Extract Week Of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week

date.head()
```

9. Extract Quarter

```python
date['quarter'] = date['date'].dt.quarter

date.head()
```

10. Extract Semester

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.head()
```

11. Extract Time elapsed between dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python
#  Calculate in diff b/w present date and real date
(today - date['date']).dt.days
```

```python
# Calculate difference in days and divide by average days in a month
np.round((today - date['date']).dt.days / 30.44, 0)
```

# Working With Time

```python
time.head()
```

```python
# Converting object to datetime datatype

time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

1. Extract Hour

```python
time['hour'] = time['date'].dt.hour

time.head()
```

2. Extract Minute

```python
time['min'] = time['date'].dt.minute

time.head()
```

3. Extract Seconds

```python
time['sec'] = time['date'].dt.second

time.head()
```

4. Extract Time Part

```python
time['time'] = time['date'].dt.time

time.head()
```

```python
time['time_year'] = time['date'].dt.year
```

```python
time['time_month'] = time['date'].dt.month
time['time_day'] = time['date'].dt.day
```

```python
time.head()
```

6. Time Differnce

```python
today - time['date']
```

```python
# Diff in Seconds

(today - time['date'])/np.timedelta64(1,'s')
```

```python
# Diff in Minutes

(today - time['date'])/np.timedelta64(1,'m')
```

```python
# Diff in Hours

(today - time['date'])/np.timedelta64(1,'h')
```

```python
# Months passed
np.round((today - time['date']).dt.days / 30.44, 0)
```
